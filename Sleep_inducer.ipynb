```python
import time
import pygame
import threading
```

```python
def play_sound(file, duration):
    pygame.mixer.init()
    pygame.mixer.music.load(file)
    pygame.mixer.music.play(-1)

    print(f"\n🎵 Playing {file} for {duration} minutes...\n")
    time.sleep(duration * 60)
    pygame.mixer.music.stop()
    print("⏹ Sound stopped. Sweet dreams 😴")
```

```python
def breathing_exercise(cycles=3):
    print("\n😌 Starting Relaxation Breathing...\n")
    for i in range(cycles):
        print("Inhale... 🌬️")
        time.sleep(4)
        print("Hold... 🤐")
        time.sleep(4)
        print("Exhale... 😮‍💨")
        time.sleep(6)
    print("\nBreathing session finished ✅\n")
```
