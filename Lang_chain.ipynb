```python
import os
import getpass

try:
    from dotenv import load_dotenv
    load_dotenv()
except ImportError:
    pass

os.environ["LANGSMITH_TRACING"] = "true"
```

```python
if "LANGSMITH_API_KEY" not in os.environ:
    os.environ["LANGSMITH_API_KEY"] = getpass.getpass(
        prompt="Enter your LangSmith API key (optional): "
    )

if "LANGSMITH_PROJECT" not in os.environ:
    project_name = getpass.getpass(
        prompt='Enter your LangSmith Project Name (default = "default"): '
    )
    os.environ["LANGSMITH_PROJECT"] = project_name if project_name else "default"

print("\nEnvironment setup complete ✅")
print("LANGSMITH_TRACING:", os.environ.get("LANGSMITH_TRACING"))
print("LANGSMITH_API_KEY:", "******" if os.environ.get("LANGSMITH_API_KEY") else "Not Provided")
print("LANGSMITH_PROJECT:", os.environ.get("LANGSMITH_PROJECT"))
```
